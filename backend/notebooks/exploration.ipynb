```python
import pandas as pd

df = pd.read_csv('../data/spam.csv', encoding='latin-1')
print(df.shape)   # combien de lignes et colonnes ?
df.head()         # afficher les 5 premières lignes
```

```python
# Garder uniquement les 2 colonnes utiles
df = df[['v1', 'v2']]

# Renommer pour que ce soit lisible
df.columns = ['label', 'message']

print(df.head())
```

```python
print(df['label'].value_counts())
print()
print(f"Pourcentage spam : {df['label'].value_counts()['spam'] / len(df) * 100:.1f}%")
```

```python
# Un exemple de spam
print("=== SPAM ===")
print(df[df['label'] == 'spam']['message'].iloc[0])
print()

# Un exemple de ham
print("=== HAM ===")
print(df[df['label'] == 'ham']['message'].iloc[0])
```

```python
df['longueur'] = df['message'].apply(len)

print(df.groupby('label')['longueur'].mean())
```
